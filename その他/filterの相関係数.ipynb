```python
import numpy as np
```

```python
import json

# JSONファイルから辞書を読み込む
with open('k1_rating0_9_filterSum.json', 'r') as f:
    filterSum09 = json.load(f)

with open('k1_rating0_5_filterSum.json', 'r') as f:
    filterSum05 = json.load(f)

with open('kyodo_filterSum.json', 'r') as f:
    filterSumkyodo = json.load(f)
```

```python
flat_kyodo = []
flat_09 = []
flat_05 = []

for key in filterSumkyodo:
    sublist_kyodo = filterSumkyodo[key]
    sublist_09 = filterSum09[key]
    sublist_05 = filterSum05[key]
    
    print(key)
    print("強度枝刈りとADMM(k1_rating = 0.9)", np.corrcoef(sublist_kyodo, sublist_09)[0, 1])
    print("強度枝刈りとADMM(k1_rating = 0.5)", np.corrcoef(sublist_kyodo, sublist_05)[0, 1])
    print("ADMM(k1_rating = 0.9)とADMM(k1_ratintg = 0.5)", np.corrcoef(sublist_09, sublist_05)[0, 1])

    flat_kyodo.extend(sublist_kyodo)
    flat_09.extend(sublist_09)
    flat_05.extend(sublist_05)

print('--------------------------------------')
print('全層の相関係数')
print("強度枝刈りとADMM(k1_rating = 0.9)", np.corrcoef(flat_kyodo, flat_09)[0, 1])
print("強度枝刈りとADMM(k1_rating = 0.5)", np.corrcoef(flat_kyodo, flat_05)[0, 1])
print("ADMM(k1_rating = 0.9)とADMM(k1_ratintg = 0.5)", np.corrcoef(flat_09, flat_05)[0, 1])
```
